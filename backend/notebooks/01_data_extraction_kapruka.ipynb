# Kapruka Data Extraction Pipeline

This notebook demonstrates how to use the KaprukaWebCrawler to automatically extract product listings across various categories on kapruka.com.

## 1. Initialization and Category Fetching
Initialize the Playwright-based crawler and retrieve all the high-level category URLs.

```python
from pprint import pprint
from src.services.web_crawler import KaprukaWebCrawler

crawler = KaprukaWebCrawler()
category_urls = crawler.get_product_category_urls()

pprint(category_urls)
```

## 2. Batch Product Scraping
Iterate through the category URLs to scrape individual product information. 


```python
for url in category_urls[11:]:
    catalog = crawler.scrape_product_details(url)
    crawler.save_to_json(catalog, output_dir='../data/catalog')
```
